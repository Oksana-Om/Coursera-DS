**Task 1.** Напишіть скрипт, який приймає рядок тексту і повертає словник, де ключами є слова, а значеннями – кількість входжень кожного слова в тексті.

````python
import string
text = 'The morning was quiet, quiet enough to hear every small sound. The trees moved gently, gently swaying in the soft wind. I walked slowly, slowly along the path, thinking about the day ahead. It felt peaceful, peaceful in a way I hadn’t experienced for a long time.'
for item in string.punctuation:
  text = text.lower().replace(item, ' ')
text_list = text.split()
words = {}
for word in text_list:
  if word in words:
    words[word] += 1
  else:
    words[word] = 1

print(words)

````


**Task 3.** Дано два списки: назви продуктів і їхні ціни.

```python
products = ["Яблуко", "Банан", "Груша"]
prices = [100, 80, 120]
```

Створіть словник, де ключами будуть назви продуктів, а значеннями — їхні ціни.


````python
products = ["Яблуко", "Банан", "Груша"]
prices = [100, 80, 120]
product_price = {}
i = 0
for product in products:
  product_price[product] = prices[i]
  i += 1
print(product_price)
````

**Task 4.** Дано список клієнтів, кожен клієнт представлений словником з полями: `"id"`, `"ім'я"`, `"вік"`, `"місто"`.

```python
[
    {"id": 1, "ім'я": "Анна", "вік": 25, "місто": "Берлін"},
    {"id": 2, "ім'я": "Іван", "вік": 30, "місто": "Берлін"},
    {"id": 3, "ім'я": "Ольга", "вік": 22, "місто": "Відень"}
]
```

Напишіть скрипт, який формує словник, у якому ключі — міста, а значення — клієнти (список) з цих міст.

````python
info = [
    {"id": 1, "ім'я": "Анна", "вік": 25, "місто": "Берлін"},
    {"id": 2, "ім'я": "Іван", "вік": 30, "місто": "Берлін"},
    {"id": 3, "ім'я": "Ольга", "вік": 22, "місто": "Відень"},
    {"id": 4, "ім'я": "Дмитро", "вік": 25, "місто": "Берлін"}
]
city_name = {}
for element in info:
  name = element["ім'я"]
  city = element["місто"]
  if city in city_name:
    city_name[city].append(name)
  else:
    city_name[city] = [name]
print(city_name)

````

**Task 5.** Дано список працівників, де кожен працівник представлений словником з полями `"ім'я"`, `"відділ"`.

```python
[
    {"ім'я": "Анна", "відділ": "Маркетинг"},
    {"ім'я": "Іван", "відділ": "Продажі"},
    {"ім'я": "Ольга", "відділ": "Маркетинг"}
]
```

Створіть скрипт, який групує працівників за відділами.

````python
list_workers = [
    {"ім'я": "Анна", "відділ": "Маркетинг"},
    {"ім'я": "Іван", "відділ": "Продажі"},
    {"ім'я": "Ольга", "відділ": "Маркетинг"}
]
depart_employee = {}
for element in list_workers:
    name = element["ім'я"]
    dep = element["відділ"]
    if dep in depart_employee:
        depart_employee[dep].append(name)
    else:
        depart_employee[dep] = [name]
print(depart_employee)
````

**Task 6.** Дано словник з даними щодо продажів, де ключі — дні місяця, а значення — сума продажу.

```python
{
    1: 500, 2: 700, 3: 300, 4: 800, 5: 200
}
```

Знайдіть дні з найбільшими відхиленнями від середнього значення.

````python
import statistics
data = {
    1: 500, 2: 700, 3: 300, 4: 800, 5: 200
}
avg = statistics.mean(data.values())
deviation_dict = {}
for i in data:
  deviation = abs(avg - data.get(i))
  if deviation in deviation_dict:
    deviation_dict[deviation].append(i)
  else:
    deviation_dict[deviation] = [i]
max_key = max(deviation_dict)
res = deviation_dict[max_key]
print(max_key, res)
````
